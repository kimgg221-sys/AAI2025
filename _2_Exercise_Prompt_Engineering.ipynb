```python
"""
Customer Support AI using a ReAct-style Prompt Chain

Stages:
1. Reason -> Understand customer problem
2. Act -> Classify issue
3. Observe -> Review classification
4. Reason -> Determine priority
5. Act -> Generate response
6. Observe -> Evaluate response quality
7. Act -> Create final support ticket
"""

# -------------------------
# Step 1: REASON
# -------------------------
def reason_about_issue(customer_message):
    reasoning = f"""
    Customer says:
    "{customer_message}"

    Analyze the customer's situation.
    Identify the core problem.
    Return a short explanation.
    """

    # Simulated AI output
    return "Customer cannot access their account because the password is not working."


# -------------------------
# Step 2: ACT
# -------------------------
def classify_issue(reasoning):
    prompt = f"""
    Based on the analysis:

    {reasoning}

    Select one category:
    - Billing
    - Technical Support
    - Account Access
    - Product Information
    """

    # Simulated AI classification
    return "Account Access"


# -------------------------
# Step 3: OBSERVE
# -------------------------
def observe_classification(category):
    observation = f"""
    Observation:
    The issue was classified as '{category}'.

    Verify whether this classification is reasonable.
    """

    return f"Classification confirmed: {category} is appropriate."


# -------------------------
# Step 4: REASON
# -------------------------
def determine_priority(reasoning, observation):
    prompt = f"""
    Analysis:
    {reasoning}

    Observation:
    {observation}

    Determine urgency:
    - High
    - Medium
    - Low
    """

    return "High"


# -------------------------
# Step 5: ACT
# -------------------------
def generate_response(category, priority):
    prompt = f"""
    Category: {category}
    Priority: {priority}

    Generate a professional customer response.
    """

    return (
        "Thank you for contacting support. "
        "We identified an account access issue and have marked it as high priority. "
        "Please attempt a password reset while our support team reviews your account."
    )


# -------------------------
# Step 6: OBSERVE
# -------------------------
def evaluate_response(response):
    prompt = f"""
    Review this response:

    {response}

    Check whether it is:
    - Professional
    - Helpful
    - Clear
    """

    return "Response passed quality check."


# -------------------------
# Step 7: ACT
# -------------------------
def create_support_ticket(
        customer_message,
        category,
        priority,
        response,
        review):

    return {
        "Customer_Message": customer_message,
        "Category": category,
        "Priority": priority,
        "Response": response,
        "Quality_Check": review,
        "Status": "Ready for Agent Review"
    }


# ==================================================
# EXECUTION LOOP (ReAct Chain)
# ==================================================

customer_message = (
    "I cannot log into my account. "
    "My password stopped working and I need access urgently."
)

# REASON
reasoning = reason_about_issue(customer_message)
print("REASONING:")
print(reasoning)

# ACT
category = classify_issue(reasoning)
print("\nACTION:")
print("Category =", category)

# OBSERVE
observation = observe_classification(category)
print("\nOBSERVATION:")
print(observation)

# REASON
priority = determine_priority(reasoning, observation)
print("\nREASONING:")
print("Priority =", priority)

# ACT
customer_response = generate_response(category, priority)
print("\nACTION:")
print(customer_response)

# OBSERVE
review = evaluate_response(customer_response)
print("\nOBSERVATION:")
print(review)

# FINAL ACTION
ticket = create_support_ticket(
    customer_message,
    category,
    priority,
    customer_response,
    review
)

print("\nFINAL SUPPORT TICKET")
print(ticket)
```
