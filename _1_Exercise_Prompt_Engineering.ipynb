```python
# Simple Customer Support AI Prompt Chain

def classify_issue(customer_message):
    prompt = f"""
    Classify the customer's issue into one category:
    - Billing
    - Technical Support
    - Account Access
    - Product Information
    - Other

    Customer Message:
    {customer_message}

    Return only the category.
    """

    # Simulated AI response
    if "password" in customer_message.lower():
        return "Account Access"
    elif "charged" in customer_message.lower() or "bill" in customer_message.lower():
        return "Billing"
    elif "error" in customer_message.lower() or "not working" in customer_message.lower():
        return "Technical Support"
    else:
        return "Other"


def determine_priority(customer_message, category):
    prompt = f"""
    Customer message:
    {customer_message}

    Category:
    {category}

    Determine priority level:
    - High
    - Medium
    - Low

    Return only the priority.
    """

    # Simulated AI response
    urgent_words = ["cannot", "unable", "locked", "urgent"]

    if any(word in customer_message.lower() for word in urgent_words):
        return "High"
    elif category == "Technical Support":
        return "Medium"
    else:
        return "Low"


def generate_customer_response(customer_message, category, priority):
    prompt = f"""
    Create a professional support response.

    Customer Message:
    {customer_message}

    Category:
    {category}

    Priority:
    {priority}

    Generate a helpful response.
    """

    return (
        f"Thank you for contacting support. "
        f"We identified your issue as '{category}'. "
        f"Your request has been assigned a {priority} priority level. "
        f"Our team will assist you as soon as possible."
    )


def create_internal_summary(customer_message, category, priority, response):
    prompt = f"""
    Create an internal support ticket summary.

    Customer Message:
    {customer_message}

    Category:
    {category}

    Priority:
    {priority}

    Customer Response:
    {response}
    """

    return {
        "Category": category,
        "Priority": priority,
        "Issue": customer_message,
        "Next Action": "Assign to appropriate support queue"
    }


# ----------------------------
# Main Workflow
# ----------------------------

customer_message = (
    "I cannot log into my account because my password isn't working."
)

# Step 1
category = classify_issue(customer_message)

# Step 2
priority = determine_priority(customer_message, category)

# Step 3
customer_response = generate_customer_response(
    customer_message,
    category,
    priority
)

# Step 4
ticket_summary = create_internal_summary(
    customer_message,
    category,
    priority,
    customer_response
)

# Output
print("Customer Response:")
print(customer_response)

print("\nInternal Ticket Summary:")
print(ticket_summary)
```
