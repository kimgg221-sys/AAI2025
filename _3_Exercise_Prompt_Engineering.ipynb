```python
"""
Customer Support AI using an Iterative ReAct Workflow
"""

customer_message = (
    "I cannot access my account and password reset is not working."
)

# =====================================================
# ITERATION 1
# =====================================================

print("=== ITERATION 1 ===")

# REASON
reasoning_1 = (
    f"Customer cannot access account. "
    f"Need to determine issue category."
)

print("REASON:", reasoning_1)

# ACT
category = "Account Access"

print("ACT: Classified issue as", category)

# OBSERVE
observation_1 = (
    f"Classification complete: {category}"
)

print("OBSERVE:", observation_1)

# =====================================================
# ITERATION 2
# =====================================================

print("\n=== ITERATION 2 ===")

# REASON
reasoning_2 = (
    f"Based on observation '{observation_1}', "
    f"determine urgency."
)

print("REASON:", reasoning_2)

# ACT
priority = "High"

print("ACT: Assigned priority =", priority)

# OBSERVE
observation_2 = (
    f"Issue is high priority because customer "
    f"cannot access account."
)

print("OBSERVE:", observation_2)

# =====================================================
# ITERATION 3
# =====================================================

print("\n=== ITERATION 3 ===")

# REASON
reasoning_3 = (
    f"Given category={category} and "
    f"priority={priority}, create response."
)

print("REASON:", reasoning_3)

# ACT
response = (
    "Your account issue has been escalated. "
    "Please stand by while we investigate."
)

print("ACT:", response)

# OBSERVE
customer_sentiment = "Customer may still need assistance"

print("OBSERVE:", customer_sentiment)

# =====================================================
# ITERATION 4
# =====================================================

print("\n=== ITERATION 4 ===")

# REASON
reasoning_4 = (
    f"Observation indicates unresolved issue. "
    f"Provide troubleshooting steps."
)

print("REASON:", reasoning_4)

# ACT
troubleshooting = (
    "Verify email address, clear browser cache, "
    "and attempt password reset again."
)

print("ACT:", troubleshooting)

# OBSERVE
observation_4 = "Additional help prepared."

print("OBSERVE:", observation_4)

# =====================================================
# FINAL ACTION
# =====================================================

support_ticket = {
    "Category": category,
    "Priority": priority,
    "Initial_Response": response,
    "Troubleshooting": troubleshooting,
    "Status": "Escalated"
}

print("\nFINAL SUPPORT TICKET")
print(support_ticket)
```
